## ConfigBox

```python
from box import ConfigBox
```

```python
d2 = ConfigBox({'key': 'value', 'key1': 'value1'})
```

```python
d2.key
```

## Pydantic BaseModel

```python
from pydantic import BaseModel
```

```python
class MyModel(BaseModel):
    key: str
    key1: str
```

```python
d3 = MyModel.model_validate({'key': 'value', 'key1': 'value1'})
```

```python
d3.key
```

# Ensure ensure_annotations v/s Pydantic validate_call

```python
def get_product(x: int, y: int) -> int:
    return x * y
```

```python
get_product(x=2, y=4)
```

```python
get_product(x=2, y='4')
```

## Ensure ensure_annotations

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def get_product(x: int, y: int) -> int:
    return x * y
```

```python
get_product(x=2, y=4)
```

```python
get_product(x=2, y='4')
```

```python
from pydantic import validate_call
```

```python
@validate_call(validate_return=True)
def get_product(x: int, y: int) -> int:
    return x * y
```

```python
get_product(x=2, y=4)
```

```python
get_product(x=2, y='4') # auto coerces
```

To disable coercion (strict mode)

```python
from pydantic import validate_call, ConfigDict
```

```python
@validate_call(config=ConfigDict(strict=True))
def get_product(x: int, y: int) -> int:
    return x * y
```

```python
get_product(x=2, y='4') # auto coerces
```
